# Match Detail Scraper

This notebook fetches HTML content from a basketball match detail page and displays the relevant tables in a tabular format.

## 1. Import Required Libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## 2. Fetch HTML Content from Match Detail URL

```python
url = 'https://vblweb.wisseq.eu/Home/MatchDetail?wedguid=BVBL25269180ORBBSW%20%20CD'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
response = requests.get(url, headers=headers)
html_content = response.text
print(f'HTTP Status: {response.status_code}')
```

## 3. Parse HTML and Extract Tables

```python
soup = BeautifulSoup(html_content, 'html.parser')
tables = soup.find_all('table')
print(f'Found {len(tables)} tables.')
```

## 4. Display All Tables in Tabular Format

```python
for idx, table in enumerate(tables):
    try:
        df = pd.read_html(str(table))[0]
        print(f'--- Table {idx+1} ---')
        display(df)
    except Exception as e:
        print(f'Could not parse table {idx+1}: {e}')
```
